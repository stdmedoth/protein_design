```python
import os
import json
import shutil

# Diretórios
cif_dir = "/kaggle/working/outputs/rfd3_designs"
pdb_dir = "/kaggle/working/outputs/rfd3_pdb"
filtered_pdb_dir = "/kaggle/working/outputs/rfd3_pdb_filtered"

# Limpa a pasta antes de começar para evitar misturar execuções antigas
shutil.rmtree(filtered_pdb_dir, ignore_errors=True)
os.makedirs(filtered_pdb_dir, exist_ok=True)

# Critérios de Aceitação Mais Rigorosos
MIN_FRACAO_ESTRUTURADA = 0.70 # Subimos a régua para 70%
MAX_CLASHES = 0
TOP_K_BACKBONES = 10

backbones_candidatos = []

print("Iniciando a filtragem rigorosa de backbones...")

for file in os.listdir(cif_dir):
    if file.endswith(".json"):
        base_name = file.replace(".json", "")
        json_path = os.path.join(cif_dir, file)
        pdb_path = os.path.join(pdb_dir, f"{base_name}.pdb")
        
        if os.path.exists(pdb_path):
            with open(json_path, 'r') as f:
                data = json.load(f)
                metrics = data.get("metrics", {})
            
            fracao = metrics.get("non_loop_fraction", 0)
            clashes = metrics.get("n_clashing.interresidue_clashes_w_backbone", 999)
            desvio = metrics.get("max_ca_deviation", 999)
            
            if fracao >= MIN_FRACAO_ESTRUTURADA and clashes <= MAX_CLASHES:
                backbones_candidatos.append({
                    "base_name": base_name,
                    "pdb_path": pdb_path,
                    "fracao": fracao,
                    "desvio": desvio
                })

# Ordenar: Maior fração estruturada primeiro, e se empatar, menor desvio do alvo
backbones_candidatos.sort(key=lambda x: (-x['fracao'], x['desvio']))

# Pegar apenas a "nata" dos backbones gerados
melhores_backbones = backbones_candidatos[:TOP_K_BACKBONES]

for b in melhores_backbones:
    destino = os.path.join(filtered_pdb_dir, f"{b['base_name']}.pdb")
    shutil.copy2(b['pdb_path'], destino)

print(f"Filtragem concluída!")
print(f"✓ Total que atingiram o limite mínimo (70%): {len(backbones_candidatos)}")
print(f"🏆 Aprovados e enviados para o MPNN (Top {TOP_K_BACKBONES}): {len(melhores_backbones)}")
```

```python
import gc
import gzip
import json
import os
import shutil
import subprocess
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
from Bio.PDB import MMCIFParser, PDBIO, PDBParser, Select
from Bio.PDB.PDBExceptions import PDBConstructionWarning

try:
    import torch
except Exception:
    torch = None

NUM_RODADAS_REFINAMENTO = int(globals().get("NUM_RODADAS_REFINAMENTO", 1))

alvo_d40g_completo = "MSYPGYPPPPGGYPPAAPGGGPWGGAAYPPPPSMPPIGLGNVATYAGQFNQDYLSGMAANMSGTFGGANMPNLYPGAPGAGYPPVPPGGFGQPPSAQQPVPPYGMYPPPGGNPPSRMPSYPPYPGAPVPGQPMPPPGQQPPGAYPGQPPVTYPGQPPVPLPGQQQPVPSYPGYPGSGTVTPAVPPTQFGSRGTITDAPGFDPLRDAEVLRKAMKGFGTDEQAIIDCLGSRSNKQRQQILLSFKTAYGKDLIKDLKSELSGNFEKTILALMKTPVLFDIYEIKEAIKGVGTDEACLIEILASRSNEHIRELNRAYKAEFKKTLEEAIRSDTSGHFQRLLISLSQGNRDESTNVDMSLAQRDAQELYAAGENRLGTDESKFNAVLCSRSRAHLVAVFNEYQRMTGRDIEKSICREMSGDLEEGMLAVVKCLKNTPAFFAERLNKAMRGAGTKDRTLIRIMVSRSETDLLDIRSEYKRMYGKSLYHDISGDTSGDYRKILLKICGGND"
fragmento_d40g = alvo_d40g_completo[24:55]

alvo_contexto = "A1-11"
select_fixed_atoms = {
    "A1-2": [],
    "A3": "N,CA,C,O,CB",
    "A4": [],
    "A5": "N,CA,C,O,CB",
    "A6": [],
    "A7": "N,CA,C,O,CB",
    "A8-10": [],
    "A11": "N,CA,C,O,CB",
}
select_hotspots = {
    "A3": "CB",
    "A5": "CB",
    "A7": "CB",
    "A11": "CB",
}

PARTIAL_T_INICIAL = 3.0
PARTIAL_T_DECAY = 0.75
PARTIAL_T_MINIMO = 1.0
TOP_K_RESULTADOS_POR_RODADA = 4

REFINEMENT_ROOT = Path("/kaggle/working/outputs/level2/refinement_rounds")
SEED_AF2_DIR = Path("/kaggle/working/outputs/af2_predictions_multimer")


def preparar_diretorio_limpo(path: Path) -> None:
    shutil.rmtree(path, ignore_errors=True)
    path.mkdir(parents=True, exist_ok=True)


def localizar_pdb_rank_001(design_dir: Path) -> Path:
    if design_dir.is_file():
        design_dir = design_dir.parent

    candidatos = sorted(design_dir.rglob("*rank_001*.pdb"))
    if not candidatos:
        candidatos = sorted(design_dir.rglob("*.pdb"))
    if not candidatos:
        raise FileNotFoundError(f"Nenhum PDB encontrado em {design_dir}")
    return candidatos[0]


def resolver_pasta_colabfold(candidate_name: str, af2_dir: Path) -> Path:
    base_name = candidate_name.split("_top")[0]

    candidate_dir = af2_dir / candidate_name
    if candidate_dir.exists():
        return candidate_dir

    base_dir = af2_dir / base_name
    if base_dir.exists():
        return base_dir

    matches = sorted(af2_dir.glob(f"{base_name}*"))
    if matches:
        return matches[0]

    return candidate_dir


def carregar_ranking_af2(af2_dir: Path, tamanho_fragmento: int) -> pd.DataFrame:
    resultados = []
    score_files = sorted(af2_dir.glob("*/*_scores_*.json"))
    if not score_files:
        raise ValueError(f"Nao foi encontrado nenhum arquivo de score em {af2_dir}")

    for score_file in score_files:
        with open(score_file, "r") as f:
            data = json.load(f)

        nome_base = score_file.name.split("_scores_")[0]
        plddt_array = data.get("plddt", [])
        protein_plddt = np.mean(plddt_array[:tamanho_fragmento]) if len(plddt_array) >= tamanho_fragmento else np.mean(plddt_array)
        binder_plddt = np.mean(plddt_array[tamanho_fragmento:]) if len(plddt_array) > tamanho_fragmento else np.mean(plddt_array)
        iptm = data.get("iptm", 0)
        ptm = data.get("ptm", 0)

        resultados.append(
            {
                "Candidato": nome_base,
                "ipTM (Afinidade)": round(iptm, 3),
                "pLDDT do Binder": round(float(binder_plddt), 1),
                "pLDDT da Proteina": round(float(protein_plddt), 1),
                "pTM Global": round(ptm, 3),
            }
        )

    df = pd.DataFrame(resultados)
    if df.empty:
        raise ValueError(f"Nao foi possivel montar o ranking a partir de {af2_dir}")

    #return df.sort_values(by=["pLDDT do Binder", "ipTM (Afinidade)"], ascending=[False, False])
    return df.sort_values(by=["ipTM (Afinidade)", "pLDDT do Binder"], ascending=[False, False])


def converter_cifs_para_pdb(cif_dir: Path, pdb_dir: Path) -> None:
    parser = MMCIFParser()
    io = PDBIO()
    for file in sorted(cif_dir.iterdir()):
        if not file.name.endswith(".cif.gz"):
            continue

        base_name = file.name.replace(".cif.gz", "")
        gz_path = file
        cif_path = cif_dir / f"{base_name}.cif"
        pdb_path = pdb_dir / f"{base_name}.pdb"

        with gzip.open(gz_path, "rb") as f_in:
            with open(cif_path, "wb") as f_out:
                f_out.write(f_in.read())

        try:
            with warnings.catch_warnings():
                warnings.simplefilter("ignore", PDBConstructionWarning)
                structure = parser.get_structure(base_name, str(cif_path))

            io.set_structure(structure)
            io.save(str(pdb_path))
            print(f"✓ Convertido com sucesso: {pdb_path.name}")
        except Exception as e:
            print(f"✗ Erro ao converter {base_name}: {e}")


def filtrar_backbones(cif_dir: Path, pdb_dir: Path, filtered_pdb_dir: Path) -> pd.DataFrame:
    min_fracao_estruturada = 0.30
    max_clashes = 0
    top_k_backbones = 10

    candidatos = []
    print("Iniciando a filtragem rigorosa de backbones...")

    for file in sorted(cif_dir.iterdir()):
        if not file.name.endswith(".json"):
            continue

        base_name = file.stem
        json_path = file
        pdb_path = pdb_dir / f"{base_name}.pdb"

        if not pdb_path.exists():
            continue

        with open(json_path, "r") as f:
            data = json.load(f)
            metrics = data.get("metrics", {})

        fracao = metrics.get("non_loop_fraction", 0)
        clashes = metrics.get("n_clashing.interresidue_clashes_w_backbone", 999)
        desvio = metrics.get("max_ca_deviation", 999)

        candidatos.append(
            {
                "base_name": base_name,
                "pdb_path": pdb_path,
                "fracao": fracao,
                "desvio": desvio,
                "clashes": clashes,
            }
        )

    candidatos_estritos = [
        c for c in candidatos
        if c["fracao"] >= min_fracao_estruturada and c["clashes"] <= max_clashes
    ]

    candidatos_estritos.sort(key=lambda x: (-x["fracao"], x["desvio"], x["clashes"]))

    if candidatos_estritos:
        melhores = candidatos_estritos[:top_k_backbones]
        modo_filtro = "estrito"
    else:
        candidatos.sort(key=lambda x: (-x["fracao"], x["desvio"], x["clashes"]))
        melhores = candidatos[:max(1, top_k_backbones)]
        modo_filtro = "fallback"

    for backbone in melhores:
        destino = filtered_pdb_dir / f"{backbone['base_name']}.pdb"
        shutil.copy2(backbone["pdb_path"], destino)

    print("Filtragem concluida!")
    print(f"✓ Total avaliados: {len(candidatos)}")
    print(f"✓ Total no filtro estrito (fracao >= 0.30 e clashes <= 0): {len(candidatos_estritos)}")
    if modo_filtro == "fallback":
        print("⚠️ Nenhum backbone passou no filtro estrito; usando fallback com os melhores disponiveis.")
    print(f"🏆 Aprovados e enviados para o MPNN (Top {top_k_backbones}): {len(melhores)}")

    return pd.DataFrame(melhores)


def gerar_mpnn(pdb_dir: Path, output_base: Path) -> None:
    output_base.mkdir(parents=True, exist_ok=True)
    pdb_files = list(sorted(pdb_dir.glob("*.pdb")))

    if not pdb_files:
        raise ValueError(f"Nenhum arquivo .pdb encontrado em {pdb_dir}")

    for p in pdb_files:
        name = p.stem
        print(f"Desenhando sequencias para o backbone: {name}")

        out_folder = output_base / name
        out_folder.mkdir(parents=True, exist_ok=True)

        cmd = [
            "python",
            "ProteinMPNN/protein_mpnn_run.py",
            "--pdb_path",
            str(p),
            "--pdb_path_chains",
            "B",
            "--out_folder",
            str(out_folder),
            "--num_seq_per_target",
            "200",
            "--sampling_temp",
            "0.25",
            "--seed",
            "42",
            "--batch_size",
            "1",
            "--suppress_print",
            "1",
        ]

        subprocess.run(cmd, check=True)


def gerar_af2_inputs(mpnn_seq_dir: Path, af2_input_dir: Path) -> int:
    top_n_por_backbone = 10
    score_maximo = 1.1

    af2_input_dir.mkdir(parents=True, exist_ok=True)
    total_geradas = 0

    print("Filtrando e formatando complexos para o AlphaFold Multimer...")

    for seq_design_folder in sorted(mpnn_seq_dir.iterdir()):
        if not seq_design_folder.is_dir():
            continue

        design_name = seq_design_folder.name
        out_subfolder = af2_input_dir / design_name
        out_subfolder.mkdir(parents=True, exist_ok=True)

        for fasta_file in sorted((seq_design_folder / "seqs").glob("*.fa")):
            base_name = fasta_file.stem

            with open(fasta_file, "r") as f:
                linhas = f.readlines()

            sequencias_avaliadas = []

            for i in range(2, len(linhas), 2):
                header = linhas[i].strip()
                seq_linha = linhas[i + 1].strip()

                try:
                    score = float(__import__("re").search(r"score=([\d\.]+)", header).group(1))
                    global_score = float(__import__("re").search(r"global_score=([\d\.]+)", header).group(1))
                except AttributeError:
                    continue

                binder_seq = seq_linha.split("/")[0]
                sequencias_avaliadas.append(
                    {
                        "binder_seq": binder_seq,
                        "score": score,
                        "global_score": global_score,
                    }
                )

            sequencias_avaliadas.sort(key=lambda x: x["global_score"])
            melhores_seqs = [seq for seq in sequencias_avaliadas if seq["score"] <= score_maximo][:top_n_por_backbone]

            contador_seq = 1
            for seq_data in melhores_seqs:
                seq_complexo = f"{fragmento_d40g}:{seq_data['binder_seq']}"
                score_formatado = f"{seq_data['score']:.3f}"
                nome_arquivo = out_subfolder / f"{base_name}_top{contador_seq}_score{score_formatado}.fasta"

                with open(nome_arquivo, "w") as out_f:
                    out_f.write(
                        f">{base_name}_top{contador_seq} | score={seq_data['score']} | global_score={seq_data['global_score']}\n{seq_complexo}\n"
                    )

                contador_seq += 1
                total_geradas += 1

    print(f"✓ Pronto! {total_geradas} complexos organizados para validacao do fragmento.")
    return total_geradas


def rodar_colabfold(af2_input_dir: Path, af2_output_dir: Path, template_dir: Path, num_models: str = "2") -> None:
    af2_output_dir.mkdir(parents=True, exist_ok=True)

    for design_path in sorted(af2_input_dir.iterdir()):
        if not design_path.is_dir():
            continue

        design_name = design_path.name
        print(f"Validando sequencias do design: {design_name}")

        design_output_dir = af2_output_dir / design_name
        design_output_dir.mkdir(parents=True, exist_ok=True)

        cmd = [
            "colabfold_batch",
            str(design_path),
            str(design_output_dir),
            "--model-type",
            "alphafold2_multimer_v3",
            "--num-recycle",
            "3",
            "--num-models",
            num_models,
            "--custom-template-path",
            str(template_dir),
            "--pair-mode",
            "unpaired_paired",
            "--stop-at-score",
            "90",
        ]

        subprocess.run(cmd, check=True)


def limpar_gpu() -> None:
    gc.collect()
    if torch is not None and torch.cuda.is_available():
        torch.cuda.empty_cache()


class CropTarget(Select):
    def accept_residue(self, residue):
        chain = residue.get_parent().id
        return 1 if chain in ["A", "B"] else 0


if NUM_RODADAS_REFINAMENTO <= 0:
    print("Nenhuma rodada de refinamento foi solicitada.")
else:
    if not SEED_AF2_DIR.exists():
        raise FileNotFoundError(f"Diretorio inicial nao encontrado: {SEED_AF2_DIR}")

    if "melhor_estrutura_inicial" in globals():
        seed_design_dir = resolver_pasta_colabfold(str(melhor_estrutura_inicial), SEED_AF2_DIR)
    else:
        seed_df = carregar_ranking_af2(SEED_AF2_DIR, tamanho_fragmento=31)
        seed_design_dir = resolver_pasta_colabfold(seed_df.iloc[0]["Candidato"], SEED_AF2_DIR)

    current_template_pdb = localizar_pdb_rank_001(seed_design_dir)
    round_summary = []
    previous_best_binder_iptm = None

    print(f"Iniciando refinamento iterativo com {NUM_RODADAS_REFINAMENTO} rodada(s)...")
    print(f"Template inicial: {current_template_pdb}")

    for round_idx in range(1, NUM_RODADAS_REFINAMENTO + 1):
        round_dir = REFINEMENT_ROOT / f"round_{round_idx:02d}"
        preparar_diretorio_limpo(round_dir)

        crop_dir = round_dir / "cropped"
        design_dir = round_dir / "rfd3_designs"
        pdb_dir = round_dir / "rfd3_pdb"
        filtered_pdb_dir = round_dir / "rfd3_pdb_filtered"
        mpnn_output_base = round_dir / "mpnn_sequences"
        af2_input_dir = round_dir / "af2_inputs_multimer"
        af2_output_dir = round_dir / "af2_predictions_multimer"

        for path in [crop_dir, design_dir, pdb_dir, filtered_pdb_dir, mpnn_output_base, af2_input_dir, af2_output_dir]:
            path.mkdir(parents=True, exist_ok=True)

        print(f"\n=== Rodada {round_idx}/{NUM_RODADAS_REFINAMENTO} ===")
        print(f"Usando como template: {current_template_pdb}")

        parser = PDBParser(QUIET=True)
        io = PDBIO()
        structure = parser.get_structure(f"round_{round_idx}", str(current_template_pdb))
        cropped_pdb_path = crop_dir / f"template_round_{round_idx:02d}.pdb"
        io.set_structure(structure)
        io.save(str(cropped_pdb_path), CropTarget())

        try:
            binder_len = len(list(structure[0]["B"].get_residues()))
        except Exception:
            raise ValueError(f"Nao foi possivel identificar a cadeia B em {current_template_pdb}")

        binder_cadeia = f"{binder_len}-{binder_len}"
        partial_t_rodada = max(PARTIAL_T_MINIMO, PARTIAL_T_INICIAL * (PARTIAL_T_DECAY ** (round_idx - 1)))
        round_config = {
            f"anexina_binder_step2_round_{round_idx:02d}": {
                "dialect": 2,
                "input": str(cropped_pdb_path),
                "contig": f"{alvo_contexto},/0,{binder_cadeia}",
                "partial_t": partial_t_rodada,
                "select_fixed_atoms": select_fixed_atoms,
                "infer_ori_strategy": "hotspots",
                "select_hotspots": select_hotspots,
                "cif_parser_args": {
                    "add_missing_atoms": True,
                },
                "is_non_loopy": True,
            }
        }

        round_json_path = round_dir / "rfd3_input_step2.json"
        with open(round_json_path, "w") as f:
            json.dump(round_config, f, indent=4)

        subprocess.run(
            [
                "bash",
                "-lc",
                f"PYTORCH_ALLOC_CONF=expandable_segments:True rfd3 design out_dir={design_dir} inputs={round_json_path} ckpt_path=/kaggle/working/models/rfd3/rfd3_latest.ckpt diffusion_batch_size=1",
            ],
            check=True,
        )

        converter_cifs_para_pdb(design_dir, pdb_dir)
        filtrar_backbones(design_dir, pdb_dir, filtered_pdb_dir)
        gerar_mpnn(filtered_pdb_dir, mpnn_output_base)
        total_geradas = gerar_af2_inputs(mpnn_output_base, af2_input_dir)

        if total_geradas == 0:
            raise ValueError(f"Nenhum complexo foi gerado na rodada {round_idx}.")

        rodar_colabfold(af2_input_dir, af2_output_dir, filtered_pdb_dir, num_models="2")

        round_df = carregar_ranking_af2(af2_output_dir, tamanho_fragmento=len(fragmento_d40g))
        top_round_df = round_df.head(TOP_K_RESULTADOS_POR_RODADA).copy()
        top_round_df.to_csv(round_dir / "top3_af2.csv", index=False)

        top_round_rows = []
        top_round_pdb_dir = round_dir / "top3_pdbs"
        top_round_pdb_dir.mkdir(parents=True, exist_ok=True)

        for rank_pos, (_, candidate_row) in enumerate(top_round_df.iterrows(), start=1):
            candidate_name = candidate_row["Candidato"]
            candidate_design_dir = resolver_pasta_colabfold(candidate_name, af2_output_dir)
            candidate_pdb = localizar_pdb_rank_001(candidate_design_dir)
            copied_candidate_pdb = top_round_pdb_dir / f"rank_{rank_pos:02d}_{candidate_name}.pdb"
            shutil.copy2(candidate_pdb, copied_candidate_pdb)
            top_round_rows.append(
                {
                    "Rodada": round_idx,
                    "Rank": rank_pos,
                    "Candidato": candidate_name,
                    "ipTM (Afinidade)": candidate_row["ipTM (Afinidade)"],
                    "pLDDT do Binder": candidate_row["pLDDT do Binder"],
                    "pLDDT da Proteina": candidate_row["pLDDT da Proteina"],
                    "pTM Global": candidate_row["pTM Global"],
                    "Template": str(candidate_pdb),
                }
            )

        best_row = top_round_df.iloc[0]
        best_name = best_row["Candidato"]
        best_design_dir = resolver_pasta_colabfold(best_name, af2_output_dir)
        best_pdb = localizar_pdb_rank_001(best_design_dir)

        round_df.to_csv(round_dir / "ranking_af2.csv", index=False)
        shutil.copy2(best_pdb, round_dir / "best_structure_round.pdb")
        round_summary.extend(top_round_rows)

        print(f"Melhor candidato da rodada {round_idx}: {best_name}")
        print(f"Template escolhido para a proxima rodada: {best_pdb}")
        
        best_binder_iptm = float(best_row["ipTM (Afinidade)"])
        #if previous_best_binder_iptm is not None and best_binder_iptm < previous_best_binder_iptm:
            #print(f"Parando: ipTM do binder caiu de {previous_best_binder_iptm:.1f} para {best_binder_iptm:.1f}.")
            #print(f"Parando: ipTM do binder caiu de {previous_best_binder_iptm:.1f} para {best_binder_iptm:.1f}.")
            #current_template_pdb = best_pdb
            #break
        
        previous_best_binder_iptm = best_binder_iptm
        current_template_pdb = best_pdb
        limpar_gpu()

    df_rodadas_refinamento = pd.DataFrame(round_summary)
    df_rodadas_refinamento = df_rodadas_refinamento.sort_values(by=["Rodada", "Rank"])
    melhor_estrutura_final = current_template_pdb

    df_rodadas_refinamento.to_csv(REFINEMENT_ROOT / "resumo_rodadas_refinamento.csv", index=False)
    shutil.copy2(melhor_estrutura_final, REFINEMENT_ROOT / "melhor_estrutura_final.pdb")

    print("\nResumo das rodadas de refinamento:")
    display(df_rodadas_refinamento)
    print(f"\nMelhor estrutura final salva em: {melhor_estrutura_final}")
```
